## 4. Bar chart: count of each property type

`value_counts()` tallies how many times each category in `property_type`
appears. `.plot(kind="bar")` turns that tally straight into a bar chart.

```
import matplotlib.pyplot as plt

counts = df["property_type"].value_counts()

counts.plot(kind="bar")
plt.title("Count of listings by property type")
plt.xlabel("Property type")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

counts
```

Look at the bars before you move on. Are any of these categories the same
thing written two different ways — "Townhome" and "Townhouse," say, or "Condo"
and "High Rise"? That's the property-type question from the assignment. Decide
now how your team wants to handle it, because it affects how you should read
this exact chart.
